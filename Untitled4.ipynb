```python
# ============================================================
# TASK 3: Improve and validate the selected model
# Drug Review Sentiment and Side-Effect Mining (UCI Drug Review)
# Compares an IMPROVED model against the Task 2 baseline on the SAME split.
# For research/education only - not medical advice.
# ============================================================
import os, re, io, sys, html, time, zipfile, warnings, subprocess
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import requests, joblib
from scipy import stats
from IPython.display import Markdown, display
from sklearn.base import clone
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import StandardScaler, label_binarize
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.model_selection import (train_test_split, GridSearchCV, StratifiedKFold, cross_val_score,
                                     cross_val_predict, GroupShuffleSplit)
from sklearn.metrics import (accuracy_score, balanced_accuracy_score, f1_score, precision_score,
                             recall_score, roc_auc_score, matthews_corrcoef, confusion_matrix)
warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")

# ---------------- SETTINGS ----------------
SELECTED_MODEL = "Linear SVM"   # your Task 2 winner: "Linear SVM" or "Logistic Regression"
OWN_FILE       = None           # path or list of paths, e.g. ["/content/drugsComTrain_raw.tsv", "/content/drugsComTest_raw.tsv"]
MAX_ROWS       = 50000          # keep the SAME value as Task 2 so the split is identical
TEST_SIZE      = 0.20
RANDOM_STATE   = 42             # keep the SAME value as Task 2
TUNE_SAMPLE    = 15000          # rows used for grid searches (final models are refit on the full train set)
CV_FOLDS       = 5
N_BOOT         = 500
LEARNING_CURVE = True           # set False to save time
TASK2_MACRO_F1 = None           # optional: type your Task 2 test macro-F1 to check the baseline reproduces it
OUT_DIR        = "results_task3"
UCI_URL = "https://archive.ics.uci.edu/static/public/462/drug+review+dataset+drugs+com.zip"
assert SELECTED_MODEL in ("Linear SVM", "Logistic Regression")
os.makedirs(OUT_DIR, exist_ok=True)

# ============ 1. LOAD DATA (same loader as Task 2) ============
def make_demo_data(n=4000, seed=0):
    """Small synthetic stand-in, ONLY used if the real dataset cannot be loaded."""
    import random
    rng = random.Random(seed)
    drugs = {"Sertraline": "Depression", "Metformin": "Diabetes", "Ortho Tri-Cyclen": "Birth Control", "Accutane": "Acne",
             "Lisinopril": "High Blood Pressure", "Zolpidem": "Insomnia", "Ibuprofen": "Pain", "Vyvanse": "ADHD"}
    se = ["nausea", "headaches", "dizziness", "weight gain", "insomnia", "dry mouth", "fatigue", "rash"]
    pos = ["This medication worked great for me, I feel so much better.", "Life changing, highly recommend it!",
           "Very happy with the results and it works fast.", "It helped a lot but I am not fully cured."]
    neg = ["This drug was terrible and did not help at all.", "Worst experience ever, I stopped taking it.",
           "I regret starting this medicine, it made things worse.", "Not helpful, I do not recommend it."]
    neu = ["It helped a little but nothing dramatic.", "Mixed feelings, some days are good and some are bad.",
           "Works okay, not great, not terrible."]
    rows = []
    for i in range(n):
        d = rng.choice(list(drugs)); k = rng.random()
        if k < .5: txt, r = rng.choice(pos), rng.randint(7, 10)
        elif k < .8: txt, r = rng.choice(neg), rng.randint(1, 4)
        else: txt, r = rng.choice(neu), rng.randint(5, 6)
        if rng.random() < .15: txt = rng.choice(pos + neg + neu)
        for _ in range(rng.randint(0, 2)):
            txt += rng.choice([f" I had {rng.choice(se)} for the first weeks.", f" No {rng.choice(se)} at all.",
                               f" I haven&#039;t had any {rng.choice(se)}."])
        rows.append({"drugName": d, "condition": drugs[d], "review": '"' + txt + '"', "rating": float(r)})
    return pd.DataFrame(rows + rows[:40])

def read_any(src, name=""):
    sep = "\t" if str(name or src).lower().endswith(".tsv") else None
    buf = io.BytesIO(src) if isinstance(src, (bytes, bytearray)) else src
    return pd.read_csv(buf, sep=sep, engine="python" if sep is None else "c")

def load_drug_reviews():
    if OWN_FILE:
        paths = [OWN_FILE] if isinstance(OWN_FILE, str) else list(OWN_FILE)
        return pd.concat([read_any(p) for p in paths], ignore_index=True)
    for u in ["https://archive.ics.uci.edu/ml/machine-learning-databases/00462/drugsCom_raw.zip", UCI_URL]:
        try:
            r = requests.get(u, timeout=120); r.raise_for_status()
            z = zipfile.ZipFile(io.BytesIO(r.content))
            parts = [read_any(z.read(n), n) for n in z.namelist() if n.lower().endswith((".tsv", ".csv"))]
            if parts:
                print("Downloaded from:", u); return pd.concat(parts, ignore_index=True)
        except Exception as ex:
            print(f"  failed: {u[:75]}... ({type(ex).__name__})")
    try:
        subprocess.run([sys.executable, "-m", "pip", "install", "-q", "ucimlrepo"], check=True)
        from ucimlrepo import fetch_ucirepo
        d = fetch_ucirepo(id=462)
        data = pd.concat([d.data.features, d.data.targets], axis=1)
        if {"review", "rating"} <= set(data.columns):
            print("Downloaded via ucimlrepo"); return data
    except Exception as ex:
        print(f"  ucimlrepo failed ({type(ex).__name__}: {str(ex)[:60]})")
    from google.colab import files
    print("\nAuto-download failed. Select drugsComTrain_raw.tsv (and drugsComTest_raw.tsv if you have it):")
    up = files.upload()
    if not up: raise RuntimeError("No file uploaded")
    return pd.concat([read_any(b, n) for n, b in up.items()], ignore_index=True)

df, USING_DEMO = None, False
try:
    df = load_drug_reviews()
except Exception as ex:
    print(f"Could not load the real dataset ({type(ex).__name__}: {str(ex)[:80]})")
    print(">>> USING A SMALL SYNTHETIC DEMO DATASET. Numbers below are NOT real results. <<<")
    df, USING_DEMO = make_demo_data(), True

std = {"drugname": "drugName", "condition": "condition", "review": "review", "rating": "rating"}
df = df.rename(columns={c: std[c.lower().strip()] for c in df.columns if c.lower().strip() in std})
df = df[[c for c in std.values() if c in df.columns]]
assert {"review", "rating"} <= set(df.columns), f"Need review + rating columns, found {list(df.columns)}"
if "drugName" not in df: df["drugName"] = "Unknown"
if len(df) > MAX_ROWS:
    df = df.sample(MAX_ROWS, random_state=RANDOM_STATE).reset_index(drop=True)
print("Loaded:", df.shape, "| demo data:", USING_DEMO)

# ============ 2. CLEAN + LABELS (identical to Task 2) ============
def clean_review(s):
    s = html.unescape(str(s))
    s = s.replace("\u2019", "'").replace("\r", " ").replace("\n", " ")
    s = re.sub(r"http\S+|www\.\S+", " ", s)
    s = s.strip().strip('"').strip("'")
    return re.sub(r"\s+", " ", s).strip()

n0 = len(df)
df["rating"] = pd.to_numeric(df["rating"], errors="coerce")
df = df.dropna(subset=["review", "rating"])
df = df[df["rating"].between(1, 10)]
df["review"] = df["review"].apply(clean_review)
df = df.drop_duplicates(subset="review")
df = df[df["review"].str.split().str.len() >= 5]
df["drugName"] = df["drugName"].fillna("Unknown").astype(str).str.strip()
df = df.reset_index(drop=True)
LABELS = ["negative", "neutral", "positive"]
df["y"] = pd.cut(df["rating"], bins=[0, 4, 6, 10], labels=LABELS).astype(str).map({l: i for i, l in enumerate(LABELS)})
print(f"Cleaning: {n0:,} -> {len(df):,} rows | class share:", df["y"].value_counts(normalize=True).sort_index().round(3).tolist())

# ============ 3. FEATURE ENGINEERING ============
# (a) Negation marking: words after "not/no/never..." become NOT_word, so "not helpful" != "helpful"
NEGATORS = {"no", "not", "never", "without", "cannot", "can't", "cant", "won't", "wont", "don't", "dont", "didn't", "didnt",
            "doesn't", "doesnt", "isn't", "isnt", "wasn't", "wasnt", "haven't", "havent", "hasn't", "hasnt", "couldn't",
            "couldnt", "wouldn't", "wouldnt", "nothing", "none", "neither", "nor"}
def mark_negation(text, window=3):
    out, left = [], 0
    for tok in re.findall(r"[a-z']+|[.,;:!?]", text.lower()):
        if tok in ".,;:!?": left = 0; continue
        t = tok.replace("'", "")
        if tok in NEGATORS: out.append(t); left = window
        elif left > 0: out.append("NOT_" + t); left -= 1
        else: out.append(t)
    return " ".join(out)

# (b) Side-effect features from a small lexicon (ties this task back to the side-effect mining goal)
LEXICON = {
    "nausea": ["nausea", "nauseous", "nauseated", "queasy"], "vomiting": ["vomiting", "vomit", "threw up"],
    "headache": ["headache", "headaches", "migraine", "migraines"], "dizziness": ["dizzy", "dizziness", "lightheaded", "vertigo"],
    "fatigue": ["fatigue", "tired", "exhausted", "drowsy", "drowsiness", "sleepy", "lethargic"],
    "insomnia": ["insomnia", "can't sleep", "couldn't sleep", "trouble sleeping"],
    "weight gain": ["weight gain", "gained weight", "gaining weight"], "weight loss": ["weight loss", "lost weight"],
    "anxiety": ["anxiety", "anxious", "panic attack", "panic attacks"], "mood": ["mood swings", "irritable", "depressed", "crying"],
    "acne": ["acne", "breakouts", "pimples"], "rash": ["rash", "hives", "itching", "itchy"],
    "diarrhea": ["diarrhea", "loose stools"], "constipation": ["constipation", "constipated"],
    "stomach": ["stomach pain", "cramps", "cramping", "bloating", "bloated"], "dry mouth": ["dry mouth", "cotton mouth"],
    "hair loss": ["hair loss", "hair falling out", "losing hair"], "sweating": ["sweating", "night sweats"],
    "tremor": ["tremor", "tremors", "shaking", "jittery"], "vision": ["blurred vision", "blurry vision"],
    "palpitations": ["palpitations", "heart racing", "racing heart"], "pain": ["muscle pain", "joint pain", "body aches"],
    "libido": ["low libido", "no sex drive", "erectile"], "brain fog": ["brain fog", "memory loss", "foggy"],
    "numbness": ["numbness", "tingling", "numb"], "swelling": ["swelling", "swollen"], "cough": ["cough", "coughing"],
}
P2C = {p: c for c, ps in LEXICON.items() for p in ps}
SE_PAT = re.compile(r"\b(?:" + "|".join(re.escape(p) for p in sorted(P2C, key=len, reverse=True)) + r")\b")
CLAUSE = re.compile(r"[.;!?]|\bbut\b|\bhowever\b|\balthough\b")
def side_effect_counts(text):
    rep, neg = set(), set()
    for clause in CLAUSE.split(text):
        for m in SE_PAT.finditer(clause):
            before = re.findall(r"[a-z']+", clause[:m.start()].split(",")[-1])[-3:]
            (neg if any(w in NEGATORS for w in before) else rep).add(P2C[m.group(0)])
    return len(rep), len(neg)

# (c) Simple text-statistics features
lc = df["review"].str.lower()
df["review_neg"] = df["review"].map(mark_negation)
words = df["review"].str.split().str.len()
df["log_words"] = np.log1p(words)
df["avg_word_len"] = df["review"].str.len() / words.clip(lower=1)
df["exclaims"] = df["review"].str.count("!")
df["questions"] = df["review"].str.count(r"\?")
df["caps_ratio"] = df["review"].apply(lambda s: sum(c.isupper() for c in s) / max(len(s), 1))
df["n_negators"] = df["review_neg"].str.split().apply(lambda t: sum(w in {n.replace("'", "") for n in NEGATORS} for w in t))
se_c = lc.map(side_effect_counts)
df["n_side_effects"] = se_c.str[0]
df["n_negated_side_effects"] = se_c.str[1]
NUM_COLS = ["log_words", "avg_word_len", "exclaims", "questions", "caps_ratio", "n_negators",
            "n_side_effects", "n_negated_side_effects"]
print("Example negation marking:", df["review_neg"].iloc[0][:110])

# ============ 4. SAME SPLIT AS TASK 2 ============
X = df[["review", "review_neg"] + NUM_COLS]
y = df["y"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=TEST_SIZE, stratify=y, random_state=RANDOM_STATE)
print(f"Train: {len(X_train):,} | Test: {len(X_test):,}")
if len(X_train) > TUNE_SAMPLE:
    Xs, _, ys, _ = train_test_split(X_train, y_train, train_size=TUNE_SAMPLE, stratify=y_train, random_state=RANDOM_STATE)
else:
    Xs, ys = X_train, y_train

# ============ 5. PIPELINE BUILDER + METRICS ============
C_GRID = [0.5, 1, 4, 10] if SELECTED_MODEL == "Logistic Regression" else [0.05, 0.1, 0.5, 1]
def make_clf(C):
    if SELECTED_MODEL == "Logistic Regression":
        return LogisticRegression(max_iter=2000, class_weight="balanced", C=C)
    return LinearSVC(class_weight="balanced", C=C)

def build(use_neg, use_char, use_num, C, word_ngram=(1, 2)):
    tfs = [("word", TfidfVectorizer(ngram_range=word_ngram, min_df=3, max_features=50000, sublinear_tf=True),
            "review_neg" if use_neg else "review")]
    if use_char:
        tfs.append(("char", TfidfVectorizer(analyzer="char_wb", ngram_range=(2, 5), min_df=5, max_features=100000,
                                            sublinear_tf=True), "review"))
    if use_num:
        tfs.append(("num", StandardScaler(), NUM_COLS))
    ct = ColumnTransformer(tfs, transformer_weights={"num": 0.3} if use_num else None)
    return Pipeline([("features", ct), ("clf", make_clf(C))])

def macro_auc(y_true, S):
    Yb = label_binarize(y_true, classes=[0, 1, 2])
    return float(np.mean([roc_auc_score(Yb[:, k], S[:, k]) for k in range(3)]))

def evaluate(y_true, pred, S):
    return {"accuracy": accuracy_score(y_true, pred), "balanced_acc": balanced_accuracy_score(y_true, pred),
            "macro_precision": precision_score(y_true, pred, average="macro"),
            "macro_recall": recall_score(y_true, pred, average="macro"),
            "macro_F1": f1_score(y_true, pred, average="macro"), "weighted_F1": f1_score(y_true, pred, average="weighted"),
            "macro_ROC_AUC": macro_auc(y_true, S), "MCC": matthews_corrcoef(y_true, pred)}

# ============ 6. BASELINE = TASK 2 MODEL (raw words, 1-2 grams, tuned C) ============
print(f"\n[Baseline] {SELECTED_MODEL}, word TF-IDF 1-2 grams on raw text")
gs_b = GridSearchCV(build(False, False, False, C_GRID[1]), {"clf__C": C_GRID}, scoring="f1_macro", cv=3, n_jobs=-1).fit(Xs, ys)
C_base = gs_b.best_params_["clf__C"]
print("Baseline tuned C =", C_base)
base = build(False, False, False, C_base)
t0 = time.time(); base.fit(X_train, y_train); base_fit_s = time.time() - t0
S_base = base.decision_function(X_test); pred_base = base.predict(X_test)
res_base = evaluate(y_test, pred_base, S_base)
print("Baseline test macro-F1: %.4f" % res_base["macro_F1"])
if TASK2_MACRO_F1 is not None:
    print("Task 2 reported %.4f -> difference %+.4f (small gaps are normal if Task 2 used a different grid)" %
          (TASK2_MACRO_F1, res_base["macro_F1"] - TASK2_MACRO_F1))

# ============ 7. FEATURE ABLATION (stage chosen by CROSS-VALIDATION on train only) ============
STAGES = [("0 Baseline: raw words", (False, False, False)),
          ("1 + negation marking", (True, False, False)),
          ("2 + char n-grams", (True, True, False)),
          ("3 + text-stat & side-effect features", (True, True, True))]
skf3 = StratifiedKFold(n_splits=3, shuffle=True, random_state=RANDOM_STATE)
abl = []
for label, flags in STAGES:
    pipe = build(*flags, C_base)
    cvs = cross_val_score(clone(pipe), X_train, y_train, cv=skf3, scoring="f1_macro", n_jobs=-1)
    m = clone(pipe).fit(X_train, y_train)
    abl.append({"stage": label, "CV_macro_F1": cvs.mean(), "CV_std": cvs.std(),
                "test_macro_F1": f1_score(y_test, m.predict(X_test), average="macro")})
    print(f"  {label:42s} CV macro-F1 {cvs.mean():.4f} (+/- {cvs.std():.4f})")
abl_df = pd.DataFrame(abl).set_index("stage")
best_stage = abl_df["CV_macro_F1"].idxmax()
best_flags = dict(STAGES)[best_stage]
print("Feature set chosen by CV:", best_stage)
display(abl_df.round(4))

# ============ 8. HYPERPARAMETER TUNING OF THE IMPROVED MODEL ============
print("\nTuning improved model (C and word n-gram range) ...")
grid = {"clf__C": C_GRID, "features__word__ngram_range": [(1, 2), (1, 3)]}
gs_i = GridSearchCV(build(*best_flags, C_base), grid, scoring="f1_macro", cv=3, n_jobs=-1).fit(Xs, ys)
print("Best params:", gs_i.best_params_, "| tuning macro-F1 %.4f" % gs_i.best_score_)
improved = clone(build(*best_flags, C_base)).set_params(**gs_i.best_params_)
t0 = time.time(); improved.fit(X_train, y_train); imp_fit_s = time.time() - t0

# Decision-threshold tuning: add class offsets to the scores, chosen on OUT-OF-FOLD train predictions
S_oof = cross_val_predict(clone(improved), X_train, y_train, cv=skf3, method="decision_function", n_jobs=-1)
y_tr_arr = np.asarray(y_train)
oof_base = f1_score(y_tr_arr, S_oof.argmax(1), average="macro")
best = (oof_base, 0.0, 0.0)
offs = np.linspace(-0.5, 0.5, 11) * S_oof.std()
for bn in offs:
    for bu in offs:
        f = f1_score(y_tr_arr, (S_oof + np.array([bn, bu, 0.0])).argmax(1), average="macro")
        if f > best[0] + 1e-9: best = (f, bn, bu)
offsets = np.array([best[1], best[2], 0.0]) if best[0] - oof_base >= 0.002 else np.zeros(3)
print("Class offsets (neg, neu, pos):", offsets.round(3), "| out-of-fold macro-F1 %.4f -> %.4f" % (oof_base, best[0]))

S_imp = improved.decision_function(X_test)
pred_imp = (S_imp + offsets).argmax(1)
res_imp = evaluate(y_test, pred_imp, S_imp)

# ============ 9. BASELINE vs IMPROVED (TEST SET) ============
comp = pd.DataFrame({"Baseline (Task 2)": res_base, "Improved (Task 3)": res_imp}).T
comp["fit_time_s"] = [base_fit_s, imp_fit_s]
delta = (comp.loc["Improved (Task 3)"] - comp.loc["Baseline (Task 2)"]).rename("Change")
print("\n=== BASELINE vs IMPROVED (held-out test set) ===")
display(pd.concat([comp, delta.to_frame().T]).round(4))

pcls = pd.DataFrame({
    "baseline_F1": f1_score(y_test, pred_base, average=None), "improved_F1": f1_score(y_test, pred_imp, average=None),
    "baseline_recall": recall_score(y_test, pred_base, average=None), "improved_recall": recall_score(y_test, pred_imp, average=None),
    "baseline_precision": precision_score(y_test, pred_base, average=None), "improved_precision": precision_score(y_test, pred_imp, average=None)},
    index=LABELS)
pcls["F1_change"] = pcls["improved_F1"] - pcls["baseline_F1"]
display(pcls.round(4))

fig, ax = plt.subplots(1, 3, figsize=(16, 4))
cm_b = confusion_matrix(y_test, pred_base, normalize="true"); cm_i = confusion_matrix(y_test, pred_imp, normalize="true")
for a, cm, t in [(ax[0], cm_b, "Baseline"), (ax[1], cm_i, "Improved")]:
    sns.heatmap(cm, annot=True, fmt=".2f", cmap="Blues", vmin=0, vmax=1, xticklabels=LABELS, yticklabels=LABELS, ax=a, cbar=False)
    a.set_title(t); a.set_xlabel("Predicted"); a.set_ylabel("True")
sns.heatmap(cm_i - cm_b, annot=True, fmt="+.2f", cmap="RdYlGn", center=0, xticklabels=LABELS, yticklabels=LABELS, ax=ax[2], cbar=False)
ax[2].set_title("Change (green on diagonal = better)"); ax[2].set_xlabel("Predicted"); ax[2].set_ylabel("True")
plt.tight_layout(); plt.savefig(f"{OUT_DIR}/confusion_before_after.png", dpi=150); plt.show()

fig, ax = plt.subplots(1, 2, figsize=(13, 4))
abl_df[["CV_macro_F1", "test_macro_F1"]].plot(kind="barh", ax=ax[0]); ax[0].set_title("Ablation: macro-F1 per feature stage")
ax[0].set_xlim(max(0, abl_df[["CV_macro_F1", "test_macro_F1"]].min().min() - 0.05), None)
pcls[["baseline_F1", "improved_F1"]].plot(kind="bar", ax=ax[1]); ax[1].set_title("Per-class F1"); ax[1].set_ylim(0, 1)
ax[1].tick_params(axis="x", rotation=0)
plt.tight_layout(); plt.savefig(f"{OUT_DIR}/ablation_and_per_class.png", dpi=150); plt.show()

# ============ 10. VALIDATION ============
print("\n=== VALIDATION ===")
# 10a. Paired k-fold cross-validation on the training set (same folds for both models)
skf = StratifiedKFold(n_splits=CV_FOLDS, shuffle=True, random_state=RANDOM_STATE)
cv_b = cross_val_score(clone(base), X_train, y_train, cv=skf, scoring="f1_macro", n_jobs=-1)
cv_i = cross_val_score(clone(improved), X_train, y_train, cv=skf, scoring="f1_macro", n_jobs=-1)
cv_diff = cv_i - cv_b
p_cv = stats.ttest_rel(cv_i, cv_b).pvalue if len(cv_diff) > 1 and cv_diff.std() > 0 else np.nan
cv_tab = pd.DataFrame({"baseline": cv_b, "improved": cv_i, "difference": cv_diff}, index=[f"fold {i+1}" for i in range(CV_FOLDS)])
cv_tab.loc["mean"] = cv_tab.mean(); cv_tab.loc["std"] = cv_tab.iloc[:CV_FOLDS].std()
print(f"10a. {CV_FOLDS}-fold CV macro-F1: baseline {cv_b.mean():.4f} vs improved {cv_i.mean():.4f}; "
      f"improved better in {int((cv_diff > 0).sum())}/{CV_FOLDS} folds; paired t-test p = {p_cv:.4f}")
display(cv_tab.round(4))

# 10b. Paired bootstrap on the test set: 95% CI of (improved - baseline) macro-F1
rng = np.random.RandomState(RANDOM_STATE); y_arr = np.asarray(y_test); diffs = []
for _ in range(N_BOOT):
    idx = rng.randint(0, len(y_arr), len(y_arr))
    diffs.append(f1_score(y_arr[idx], pred_imp[idx], average="macro") - f1_score(y_arr[idx], pred_base[idx], average="macro"))
diffs = np.array(diffs); d_lo, d_hi = np.percentile(diffs, [2.5, 97.5])
print(f"10b. Bootstrap difference in macro-F1: {diffs.mean():+.4f} (95% CI {d_lo:+.4f} to {d_hi:+.4f})")

# 10c. McNemar test on test-set errors
b_only = int(((pred_base == y_arr) & (pred_imp != y_arr)).sum())   # baseline right, improved wrong
i_only = int(((pred_base != y_arr) & (pred_imp == y_arr)).sum())   # baseline wrong, improved right
p_mc = stats.binomtest(min(b_only, i_only), b_only + i_only, 0.5).pvalue if b_only + i_only > 0 else 1.0
print(f"10c. McNemar: improved fixed {i_only} reviews, broke {b_only}; p = {p_mc:.4f}")

# 10d. Unseen drugs
gss = GroupShuffleSplit(n_splits=1, test_size=TEST_SIZE, random_state=RANDOM_STATE)
g_tr, g_te = next(gss.split(X, y, groups=df["drugName"]))
gb = clone(base).fit(X.iloc[g_tr], y.iloc[g_tr]); gi = clone(improved).fit(X.iloc[g_tr], y.iloc[g_tr])
g_b = f1_score(y.iloc[g_te], gb.predict(X.iloc[g_te]), average="macro")
g_i = f1_score(y.iloc[g_te], gi.predict(X.iloc[g_te]), average="macro")
print(f"10d. Unseen-drug macro-F1: baseline {g_b:.4f} vs improved {g_i:.4f} ({g_i - g_b:+.4f})")

# 10e. Learning curve
lc_df = None
if LEARNING_CURVE:
    rows = []
    for frac in [0.25, 0.5, 0.75, 1.0]:
        if frac < 1.0:
            Xa, _, ya, _ = train_test_split(X_train, y_train, train_size=frac, stratify=y_train, random_state=RANDOM_STATE)
        else:
            Xa, ya = X_train, y_train
        rows.append({"train_fraction": frac, "n_train": len(Xa),
                     "baseline": f1_score(y_test, clone(base).fit(Xa, ya).predict(X_test), average="macro"),
                     "improved": f1_score(y_test, clone(improved).fit(Xa, ya).predict(X_test), average="macro")})
    lc_df = pd.DataFrame(rows).set_index("train_fraction")
    ax = lc_df[["baseline", "improved"]].plot(marker="o", figsize=(6, 3.8)); ax.set_ylabel("test macro-F1"); ax.set_title("Learning curve")
    plt.tight_layout(); plt.savefig(f"{OUT_DIR}/learning_curve.png", dpi=150); plt.show()
    print("10e. Learning curve (does more data still help?):"); display(lc_df.round(4))

# ============ 11. ERROR ANALYSIS + WHAT THE NEW FEATURES DO ============
pd.set_option("display.max_colwidth", 150)
fixed = (pred_base != y_arr) & (pred_imp == y_arr); broken = (pred_base == y_arr) & (pred_imp != y_arr)
ex = pd.DataFrame({"review": X_test["review"].values, "true": [LABELS[i] for i in y_arr],
                   "baseline": [LABELS[i] for i in pred_base], "improved": [LABELS[i] for i in pred_imp]})
print(f"\nImproved model fixed {fixed.sum()} test reviews and broke {broken.sum()}.")
print("Examples FIXED by the improved model:"); display(ex[fixed].head(5))
print("Examples BROKEN by the improved model:"); display(ex[broken].head(5))

feat_names = improved.named_steps["features"].get_feature_names_out()
coef = improved.named_steps["clf"].coef_
num_idx = [i for i, n in enumerate(feat_names) if n.startswith("num__")]
if num_idx:
    print("\nWeights of the engineered numeric features (per class, larger = pushes toward that class):")
    display(pd.DataFrame(coef[:, num_idx].T, index=[feat_names[i][5:] for i in num_idx], columns=LABELS).round(3))
word_idx = [i for i, n in enumerate(feat_names) if n.startswith("word__") and "not_" in n.lower()]
if word_idx:
    top = sorted(word_idx, key=lambda i: -abs(coef[:, i]).max())[:10]
    print("Strongest negation-marked terms:", ", ".join(feat_names[i][6:] for i in top))

# ============ 12. WRITTEN FINDINGS (generated from YOUR numbers) ============
d_f1 = res_imp["macro_F1"] - res_base["macro_F1"]
rel = 100 * d_f1 / max(res_base["macro_F1"], 1e-9)
sig_boot = d_lo > 0
if sig_boot and cv_diff.mean() > 0:
    verdict = "The improvement is **statistically meaningful**: the bootstrap interval excludes 0 and cross-validation agrees."
elif d_f1 > 0:
    verdict = "The improvement is **small and not conclusively significant** (the bootstrap interval includes 0), so treat it as a modest gain."
else:
    verdict = "**No real improvement was found**; the baseline is as good, so the simpler model should be kept."
best_cls = pcls["F1_change"].idxmax(); worst_cls = pcls["F1_change"].idxmin()
step_gain = abl_df["CV_macro_F1"].diff().dropna()
biggest = step_gain.idxmax() if len(step_gain) else "n/a"
findings = f"""
### Findings: baseline vs improved {SELECTED_MODEL}

**Headline.** Test macro-F1 moved from **{res_base['macro_F1']:.3f}** to **{res_imp['macro_F1']:.3f}** ({d_f1:+.3f}, {rel:+.1f}% relative); accuracy {res_base['accuracy']:.3f} to {res_imp['accuracy']:.3f}, macro ROC-AUC {res_base['macro_ROC_AUC']:.3f} to {res_imp['macro_ROC_AUC']:.3f}, MCC {res_base['MCC']:.3f} to {res_imp['MCC']:.3f}. {verdict}

**What was changed.** Feature set chosen by cross-validation: *{best_stage}*. Hyperparameters: {gs_i.best_params_}. Class-score offsets (neg, neu, pos): {np.round(offsets, 3).tolist()}.

**Evidence.**
- Paired {CV_FOLDS}-fold CV macro-F1: {cv_b.mean():.3f} vs {cv_i.mean():.3f}; improved wins {int((cv_diff > 0).sum())}/{CV_FOLDS} folds (paired t-test p = {p_cv:.3f}).
- Paired bootstrap on the test set: difference {diffs.mean():+.4f}, 95% CI [{d_lo:+.4f}, {d_hi:+.4f}].
- McNemar: improved fixed {i_only} reviews and broke {b_only} (p = {p_mc:.3f}).
- Unseen drugs: {g_b:.3f} vs {g_i:.3f} ({g_i - g_b:+.3f}), so the gain {'holds' if g_i > g_b else 'does not hold'} on drugs not seen in training.
- Per class, F1 changed most for **{best_cls}** ({pcls.loc[best_cls, 'F1_change']:+.3f}) and least for **{worst_cls}** ({pcls.loc[worst_cls, 'F1_change']:+.3f}).
- Largest single stage gain in the ablation (CV): **{biggest}**.

**Cost.** Training time {base_fit_s:.1f}s to {imp_fit_s:.1f}s; features per model: {len(feat_names):,} (improved).

**Interpretation.** Negation marking helps because phrases like "not helpful" or "no relief" now have their own features instead of being confused with their positive counterparts. Character n-grams help with misspellings and informal wording in patient text. The text-statistic and side-effect features add a little extra signal but are weak on their own, since a patient can mention side effects and still rate the drug highly. The neutral class (ratings 5-6) remains the hardest: its wording overlaps with both neighbours, and the ratings are only a noisy proxy for sentiment.

**Limitations.** Gains are measured against labels derived from ratings; the lexicon covers only listed side effects; CV compares the models without the class offsets (offsets are evaluated on the test set only); results come from {len(df):,} reviews.
"""
if USING_DEMO: findings = "**WARNING: SYNTHETIC demo data. These numbers are not real results.**\n" + findings
display(Markdown(findings))

# ============ 13. SAVE ============
pd.concat([comp, delta.to_frame().T]).round(4).to_csv(f"{OUT_DIR}/baseline_vs_improved.csv")
pcls.round(4).to_csv(f"{OUT_DIR}/per_class_comparison.csv")
abl_df.round(4).to_csv(f"{OUT_DIR}/feature_ablation.csv")
cv_tab.round(4).to_csv(f"{OUT_DIR}/paired_cv.csv")
if lc_df is not None: lc_df.round(4).to_csv(f"{OUT_DIR}/learning_curve.csv")
joblib.dump({"pipeline": improved, "class_offsets": offsets, "labels": LABELS}, f"{OUT_DIR}/improved_model.joblib")
with open(f"{OUT_DIR}/task3_findings.md", "w") as f:
    f.write("# Task 3: Improvement and validation\n\n" + findings + "\n\n## Test-set comparison\n\n" +
            pd.concat([comp, delta.to_frame().T]).round(4).to_markdown() + "\n\n## Feature ablation\n\n" +
            abl_df.round(4).to_markdown() + "\n\n## Paired cross-validation\n\n" + cv_tab.round(4).to_markdown() + "\n")
print("Saved to ./%s:" % OUT_DIR, sorted(os.listdir(OUT_DIR)))
try:
    import shutil; shutil.make_archive("task3_results", "zip", OUT_DIR)
    from google.colab import files; files.download("task3_results.zip")
except Exception:
    pass
```
